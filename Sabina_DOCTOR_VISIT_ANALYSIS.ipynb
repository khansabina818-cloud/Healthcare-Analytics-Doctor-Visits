```python
#visualize the percentage of people getting govt health insurance due to low income, due to old age and also the percentage of the people having private health
```

```python
# Visualize the percentage of people getting govt health insurance
# due to low income, due to old age, and having private health insurance

label = ['yes', 'no']


# 1. Govt health insurance due to low income
Y = df[df['freepoor'] == 'yes']
N = df[df['freepoor'] == 'no']

x = [Y.shape[0], N.shape[0]]

plt.figure(figsize=(5,5))
plt.pie(x, labels=label, autopct='%1.1f%%')
plt.title("% of people getting govt health insurance due to low income")
plt.show()


# 2. Govt health insurance due to old age, disability or veteran status
Y = df[df['freerepat'] == 'yes']
N = df[df['freerepat'] == 'no']

x = [Y.shape[0], N.shape[0]]

plt.figure(figsize=(5,5))
plt.pie(x, labels=label, autopct='%1.1f%%')
plt.title("% of people getting govt health insurance due to old age, disability or veteran status")
plt.show()


# 3. People having private health insurance
Y = df[df['private'] == 'yes']
N = df[df['private'] == 'no']

x = [Y.shape[0], N.shape[0]]

plt.figure(figsize=(5,5))
plt.pie(x, labels=label, autopct='%1.1f%%')
plt.title("% of people having private health insurance")
plt.show()
```

```python
#plot a horizontal bar chart to abalyse the reduced days of activity due to illness based on Gender
```

```python
db = df.groupby('gender')['reduced'].sum().to_frame().reset_index()

#Creating the bar chart
plt.barh(db['gender'], db['reduced'], color=['cornflowerblue', 'lightseagreen'])

#Adding the aesthetics
plt.title('Bar Chart')
plt.xlabel('reduced activity')
plt.ylabel('gender')

#Show the plot
plt.show()
```
